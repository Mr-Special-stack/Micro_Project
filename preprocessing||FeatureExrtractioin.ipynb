### Handling Missing Values

I will impute missing values in numerical columns using the mean of each column. This is a common strategy when the number of missing values is not excessively large, or as a first approach for larger gaps.

```python
# Impute missing numerical values with the mean
for col in df.columns:
    if df[col].dtype in ['float64', 'int64']:
        if df[col].isnull().any():
            df[col] = df[col].fillna(df[col].mean())

# Verify that there are no more missing values
missing_values_after_imputation = df.isnull().sum()
display(missing_values_after_imputation[missing_values_after_imputation > 0])
```

### Feature Extraction: One-Hot Encoding for 'State'

The 'State' column is a categorical variable. To use it in most machine learning models, I will convert it into a numerical format using one-hot encoding.

```python
# Perform one-hot encoding on the 'State' column
df = pd.get_dummies(df, columns=['State'], drop_first=True)

display(df.head())
```

### Outlier Detection in `corr_daily_energy_met_y`

I will use the Interquartile Range (IQR) method to detect outliers in the target variable `corr_daily_energy_met_y`. This method identifies values that fall below `Q1 - 1.5 * IQR` or above `Q3 + 1.5 * IQR`.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate IQR for 'corr_daily_energy_met_y'
Q1 = df['corr_daily_energy_met_y'].quantile(0.25)
Q3 = df['corr_daily_energy_met_y'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify outliers
outliers = df[(df['corr_daily_energy_met_y'] < lower_bound) | (df['corr_daily_energy_met_y'] > upper_bound)]

print(f"Number of outliers in 'corr_daily_energy_met_y': {len(outliers)}")
print(f"Lower bound for outliers: {lower_bound:.2f}")
print(f"Upper bound for outliers: {upper_bound:.2f}")

# Visualize outliers using a box plot
fig = plt.figure(figsize=(10, 6))
sns.boxplot(y=df['corr_daily_energy_met_y'])
plt.title('Box Plot of Corrected Daily Energy Consumption (with Outliers)')
plt.ylabel('Corrected Daily Energy Consumption (MU)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

### Outlier Handling: Winsorization

To handle the detected outliers without removing a significant portion of data, I will apply winsorization. This technique caps the outlier values at the defined upper and lower bounds (Q1 - 1.5 * IQR and Q3 + 1.5 * IQR) rather than removing them.

```python
# Apply winsorization (clipping) to 'corr_daily_energy_met_y'
df['corr_daily_energy_met_y'] = df['corr_daily_energy_met_y'].clip(lower=lower_bound, upper=upper_bound)

# Verify the change by re-plotting the box plot
fig = plt.figure(figsize=(10, 6))
sns.boxplot(y=df['corr_daily_energy_met_y'])
plt.title('Box Plot of Corrected Daily Energy Consumption (after Winsorization)')
plt.ylabel('Corrected Daily Energy Consumption (MU)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()

# Optionally, you can also check descriptive statistics after clipping
display(df['corr_daily_energy_met_y'].describe())
```

### Preparing the Dataset for Training

Now that outliers have been handled, I will separate the features (X) from the target variable (y) and then apply standard scaling to the numerical features to ensure they are on a similar scale. This is a crucial step for many machine learning algorithms.

```python
from sklearn.preprocessing import StandardScaler

# Define features (X) and target (y)
X = df.drop('corr_daily_energy_met_y', axis=1)
y = df['corr_daily_energy_met_y']

# Identify numerical columns for scaling (excluding 'year', 'month', 'day' if not desired to be scaled,
# but typically they are scaled too for most models)
# Boolean columns (from one-hot encoding) should not be scaled
numerical_cols = X.select_dtypes(include=['float64', 'int32']).columns

# Initialize StandardScaler
scaler = StandardScaler()

# Apply scaling to numerical features
X[numerical_cols] = scaler.fit_transform(X[numerical_cols])

display(X.head())
display(y.head())
```
